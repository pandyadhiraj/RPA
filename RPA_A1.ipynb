```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
```

```python
# Assuming the file name is 'animal_shelter_data.csv'
df = pd.read_csv('/content/drive/MyDrive/Animal_Center_Intakes.csv')

# Display the first few rows of the dataset
df.head()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# Frequency count of categorical variables
df['Intake Type'].value_counts()
df['Animal Type'].value_counts()
df['Breed'].value_counts()
```

```python
# Plotting a histogram for Age upon Intake (if it's numeric)
df['Age upon Intake'].value_counts().sort_index().plot(kind='bar', figsize=(10,6))
plt.title('Age upon Intake Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# Plotting bar plot for Animal Type
sns.countplot(data=df, x='Animal Type', palette='Set2')
plt.title('Distribution of Animal Types')
plt.xlabel('Animal Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()

# Plotting bar plot for Breed
plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='Breed', order=df['Breed'].value_counts().index, palette='Set3')
plt.title('Distribution of Animal Breeds')
plt.xlabel('Count')
plt.ylabel('Breed')
plt.show()
```

```python
# Boxplot for Age upon Intake across different Animal Types
sns.boxplot(data=df, x='Animal Type', y='Age upon Intake')
plt.title('Age Distribution by Animal Type')
plt.xlabel('Animal Type')
plt.ylabel('Age upon Intake')
plt.xticks(rotation=45)
plt.show()
```

```python
# Boxplot for Age upon Intake across different Animal Types
sns.boxplot(data=df, x='Animal Type', y='Age upon Intake')
plt.title('Age Distribution by Animal Type')
plt.xlabel('Animal Type')
plt.ylabel('Age upon Intake')
plt.xticks(rotation=45)
plt.show()
```

```python
# If there were other numerical columns, you could check for correlations
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
df['Month'] = df['DateTime'].dt.month
df['Year'] = df['DateTime'].dt.year

# Plotting the number of animals admitted by month
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Month', palette='viridis')
plt.title('Animals Intake by Month')
plt.xlabel('Month')
plt.ylabel('Count')
plt.xticks(ticks=np.arange(0, 12), labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.show()
```

```python
# Plotting the number of animals admitted over the years
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Year', palette='magma')
plt.title('Animals Intake by Year')
plt.xlabel('Year')
plt.ylabel('Count')
plt.show()
```

```python
# If there are more numerical columns
sns.pairplot(df, hue='Animal Type')
plt.show()
```

```python
# Checking for outliers in Age upon Intake
sns.boxplot(data=df, x='Age upon Intake')
plt.title('Age upon Intake Boxplot')
plt.show()
```

```python
# Plotting a histogram and KDE for Age upon Intake
plt.figure(figsize=(10, 6))
sns.histplot(df['Age upon Intake'].str.extract('(\d+)').astype(float), kde=True, bins=20, color='skyblue')
plt.title('Age upon Intake Distribution (with KDE)')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Stacked bar plot for Intake Type vs Animal Type
plt.figure(figsize=(12, 6))
pd.crosstab(df['Animal Type'], df['Intake Type']).plot(kind='bar', stacked=True, figsize=(12, 6), colormap='Set1')
plt.title('Intake Type Distribution by Animal Type')
plt.xlabel('Animal Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# Gender distribution across Animal Type
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Animal Type', hue='Gender upon Intake', palette='Set2')
plt.title('Gender Distribution by Animal Type')
plt.xlabel('Animal Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# Pie chart for Intake Condition
plt.figure(figsize=(8, 8))
df['Intake Condition'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=sns.color_palette('Set2'))
plt.title('Distribution of Intake Conditions')
plt.ylabel('')
plt.show()
```
